```python
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_excel("Ev_market_Analysis.xlsx", sheet_name="process_data")
```

Price vs Horsepower

```python
plt.scatter(df['horsepower'], df['price_usd'])

plt.xlabel("Horsepower")
plt.ylabel("Price USD")
plt.title("Price vs Horsepower")

plt.show()
```

Higher horsepower EVs generally command premium prices.

Range vs Customer Rating

```python
plt.scatter(df['range_miles'], df['customer_rating'])

plt.xlabel("Range Miles")
plt.ylabel("Customer Rating")
plt.title("Range vs Customer Rating")

plt.show()
```

EVs with longer driving ranges generally receive higher customer ratings. 
This suggests that driving range is one of the most important factors influencing customer satisfaction in the EV market.

Autopilot Level vs Price

```python
plt.boxplot(
    [df[df['autopilot_level']==0]['price_usd'],
     df[df['autopilot_level']==1]['price_usd'],
     df[df['autopilot_level']==2]['price_usd'],
     df[df['autopilot_level']==3]['price_usd']]
)

plt.xlabel("Autopilot Level")
plt.ylabel("Price USD")
plt.title("Autopilot Level vs Price")

plt.show()
```

Vehicles with higher autopilot levels tend to have higher prices. 
This indicates that advanced autonomous technology is positioned as a premium feature in modern EVs.

Performance vs Cost Efficiency

```python
plt.scatter(df['Performance_Score'], df['Cost_per_mile'])

plt.xlabel("Performance Score")
plt.ylabel("Cost Per Mile")
plt.title("Performance vs Cost Efficiency")

plt.show()
```

High-performance EVs usually show higher cost per mile values. 
This suggests that increased performance often comes at the expense of cost efficiency and affordability.
